```python
import matplotlib.pyplot as plt
import matplotlib as mpl

# Enable LaTeX rendering
mpl.rcParams['text.usetex'] = True
mpl.rcParams['font.family'] = 'serif'

fig, ax1 = plt.subplots(figsize=(15, 9), facecolor='white')

ax2 = ax1.twinx()
ax2.plot(h, s22, '--ko', linewidth=2)
ax1.plot(h, sif_md, '-kd', linewidth=2)


ax1.set_xlabel(r'Normalized crack depth from the loading boundary', fontsize=32)
ax2.set_ylabel(r'$\sigma_{22\,\mathrm{(boundary)}}$ (GPa)', fontsize=32)
ax1.set_ylabel(r'$K_{IC}$ (MPa $\sqrt{\mathrm{m}}$)', fontsize=32)  # Fixed extra } brace

ax2.annotate(
    r'corresponding value of $K_I$' + '\n' + r'($R^2$ (\%))',
    xy=(0.7825, 1.9), xytext=(0.5, 1), fontsize=32,
    arrowprops=dict(facecolor='gray', shrink=0.05)
)
ax2.annotate(
    r'critical boundary stress' + '\n' + r'before crack propagation',
    xy=(0.715, 2.5), xytext=(0.775, 2.75), fontsize=32,
    arrowprops=dict(facecolor='gray', shrink=0.05)
)

annotations = [
    (0.48,   3.125, '(99.43)'),
    (0.5,    3.34,  '(99.38)'),
    (0.525,  3.13,  '(99.42)'),
    (0.5525, 3.34,  '(99.42)'),
    (0.60,   3.34,  '(99.36)'),
    (0.615,  2.75,  '(98.51)'),
    (0.665,  2.25,  '(99.32)'),
    (0.72,   1.975, '(99.40)'),
    (0.775,  1.58,  '(99.95)'),
    (0.852,  1.55,  '(99.68)'),
    (0.90,   1.20,  '(99.72)'),
    (0.91,   0.845, '(99.04)'),
]

for x, y, label in annotations:
    ax2.annotate(label, xy=(x, y), xytext=(x, y), fontsize=21, color='black')

plt.rc('axes',  titlesize=24)
plt.rc('xtick', labelsize=24)
plt.rc('ytick', labelsize=24)
plt.rc('axes',  labelsize=24)

labels = [r'', r'$h/2$', r'$h/2.5$', r'$h/3.33$', r'$h/5$', r'$h/10$']
ax1.set_xticklabels(labels)

plt.savefig('fig_8.pdf', dpi=1200, bbox_inches='tight')
```
