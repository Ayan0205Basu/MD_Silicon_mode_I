```python
import matplotlib as mpl
mpl.rcParams['font.family']='serif'
mpl.rcParams['mathtext.fontset']='cm'
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.pyplot import legend,gca
from mpl_toolkits.axes_grid1.inset_locator import zoomed_inset_axes
from mpl_toolkits.axes_grid1.inset_locator import mark_inset
```

```python
dt = pd.read_excel (r'stress_data_75.xlsx')
mat = dt.to_numpy()
new_mat=np.transpose(mat)
```

```python
pos_x = new_mat[1]
pos_x = pos_x/10
pos_x
```

```python
sig_yy =new_mat[11]
sig_yy
```

```python
fig,ax = plt.subplots(figsize=(15,9),facecolor = 'white')
#plt.subplot(1,2,1)
plotss = plt.plot(pos_x,sig_yy,'--ko', linewidth=1, label = "$\sigma_{yy}$ curve obtained from MD")
plt.xlabel('X co-ordinate of atom position (nm)')
plt.ylabel('$\sigma_{yy}$ (GPa)')
#plt.xlim([0.66,0.75])
#plt.ylim([0, 13])
plt.rc('axes', titlesize=18) 
plt.rc('xtick', labelsize=18) 
plt.rc('ytick', labelsize=18) 
plt.rc('axes', labelsize=18)

#ax.legend()
plt.legend(loc=1, prop={'size': 18})
plt.legend(loc=1, prop={'size': 18})
plt.title('')
#grids
plt.minorticks_on()
#plt.grid (which='minor', color='g', ls='-.', lw=0.5)
#plt.grid(which='major',color='g', ls='--', lw=0.5)
#################for zooming particular range


#plt.show()
plt.savefig('mid_75_s_local.pdf',dpi = 1200)
```

```python
boundary_stress = 2.247614738
total_width = 217.2
```

```python
normalized_stress = sig_yy/boundary_stress
normalized_x = pos_x/total_width
```

```python
fig,ax = plt.subplots(figsize=(15,9),facecolor = 'white')
#plt.subplot(1,2,1)
plotss = plt.plot(normalized_x,normalized_stress,'--ko', linewidth=1, label = "normalized data for edge crack")
plt.xlabel('$x/W$')
plt.ylabel('$\sigma_{yy}/\sigma_{yy(boundary)}$')
#plt.xlim([0.66,0.75])
#plt.ylim([0, 13])
plt.rc('axes', titlesize=18) 
plt.rc('xtick', labelsize=18) 
plt.rc('ytick', labelsize=18) 
plt.rc('axes', labelsize=18)

#ax.legend()
plt.legend(loc=1, prop={'size': 18})
plt.legend(loc=1, prop={'size': 18})
plt.title('')
#grids
plt.minorticks_on()
#plt.grid (which='minor', color='g', ls='-.', lw=0.5)
#plt.grid(which='major',color='g', ls='--', lw=0.5)
#################for zooming particular range


#plt.show()
plt.savefig('mid_80_s_local_normalized.pdf',dpi = 1200)
```

```python
xx = pos_x-(2*total_width/3)
xx
```

```python
stresses_mpa = sig_yy*1000
xx_m = xx*10**(-9)
stresses_mpa
```

```python
Y_inv = np.sqrt(2*np.pi*xx_m)
YY = 1/Y_inv
YY
```

```python
#remove the entry of the first atom
YY = YY[1:]
stresses_mpa = stresses_mpa[1:]
```

```python
fig,ax = plt.subplots(figsize=(15,9),facecolor = 'white')
#plt.subplot(1,2,1)
plotss = plt.plot(YY,stresses_mpa,'--ko', linewidth=1, label = "normalized data for edge crack")
plt.xlabel('$x/W$')
plt.ylabel('$\sigma_{yy}/\sigma_{yy(boundary)}$')
#plt.xlim([0.66,0.75])
#plt.ylim([0, 13])
plt.rc('axes', titlesize=18) 
plt.rc('xtick', labelsize=18) 
plt.rc('ytick', labelsize=18) 
plt.rc('axes', labelsize=18)

#ax.legend()
plt.legend(loc=1, prop={'size': 18})
plt.legend(loc=1, prop={'size': 18})
plt.title('')
#grids
plt.minorticks_on()
#plt.grid (which='minor', color='g', ls='-.', lw=0.5)
#plt.grid(which='major',color='g', ls='--', lw=0.5)
#################for zooming particular range


plt.show()
#plt.savefig('edge_s_local_normalized.pdf',dpi = 1200)
```

```python
ss_cut =stresses_mpa[4:5]
y_cut = YY[4:5]
```

```python
m_x = np.mean(y_cut)
m_y = np.mean(ss_cut)
kic = m_y/m_x
kic
```

```python
b_0 = 0     
b_1 = 0.9307963119707106
```

```python
stress_fit = b_1 * y_cut + b_0 
error = ss_cut - stress_fit
n_1 = np.size(y_cut)
ss_mean = np.mean(ss_cut)
se = np.sum(error**2)
print('squared error is', se)
mse = se/n_1 
print('mean squared error is', mse)
SSt = np.sum((ss_cut - ss_mean)**2)
R2 = 1- (se/SSt)
print('R square is', R2)
```

```python

for i in range(2,11):
    print(i+1)
    ss_cut =stresses_mpa[0:i]
    y_cut = YY[0:i]

    m_x = np.mean(y_cut)
    m_y = np.mean(ss_cut)
    kic = m_y/m_x
    print(kic)

    b_0 = 0     
    b_1 = kic

    stress_fit = b_1 * y_cut + b_0 
    error = ss_cut - stress_fit
    n_1 = np.size(y_cut)
    ss_mean = np.mean(ss_cut)
    se = np.sum(error**2)
    print('squared error is', se)
    mse = se/n_1 
    print('mean squared error is', mse)
    SSt = np.sum((ss_cut - ss_mean)**2)
    R2 = 1- (se/SSt)
    print('R square is', R2)
```
