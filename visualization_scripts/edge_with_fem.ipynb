```python
import matplotlib as mpl
mpl.rcParams['font.family']='serif'
mpl.rcParams['mathtext.fontset']='cm'
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.pyplot import legend,gca
from mpl_toolkits.axes_grid1.inset_locator import zoomed_inset_axes
from mpl_toolkits.axes_grid1.inset_locator import mark_inset
```

```python
dt = pd.read_excel (r'nd_edge_604.xlsx')
mat = dt.to_numpy()
new_mat=np.transpose(mat)

pos_norm_604 = new_mat[1]


sig_norm_604 =new_mat[0]
```

```python
dt = pd.read_excel (r'nd_edge_5492.xlsx')
mat = dt.to_numpy()
new_mat=np.transpose(mat)

pos_norm_5492 = new_mat[1]


sig_norm_5492 =new_mat[0]
```

```python
dt = pd.read_excel (r'nd_edge_45588.xlsx')
mat = dt.to_numpy()
new_mat=np.transpose(mat)

pos_norm_45588 = new_mat[1]


sig_norm_45588 =new_mat[0]
```

```python
dt = pd.read_excel (r'edge_stress_data.xlsx')
mat = dt.to_numpy()
new_mat=np.transpose(mat)


pos_x = new_mat[1]
pos_x = pos_x/10
pos_x

sig_yy =new_mat[11]
sig_yy

boundary_stress = 2.201778333
total_width = 217.2

normalized_stress = sig_yy/boundary_stress
normalized_x = pos_x/total_width
normalized_stress,normalized_x

fig,ax = plt.subplots(figsize=(15,9),facecolor = 'white')
#plt.subplot(1,2,1)
plotss = plt.plot(normalized_x,normalized_stress,'-ro', linewidth=1, label = "virial stress analysis")
plotss = plt.plot(pos_norm_604,sig_norm_604,'--kd', linewidth=1, label = "604 element FE analysis")
plotss = plt.plot(pos_norm_5492,sig_norm_5492,'-.k*', linewidth=1, label = "5492 element FE analysis")
plotss = plt.plot(pos_norm_45588,sig_norm_45588,'--kv', linewidth=1, label = "45588 element FE analysis")
ax.annotate('1$^{\mathrm{st}}$ atomic site', xy=(0.336, 11.10688302), xytext=(0.366, 12.10688302), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
ax.annotate('2$^{\mathrm{nd}}$ atomic site', xy=(0.3385, 8.75518363), xytext=(0.3685, 9.75518363), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
ax.annotate('3$^{\mathrm{rd}}$ atomic site', xy=(0.341, 6.38771691), xytext=(0.371, 7.38771691), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
ax.annotate('4$^{\mathrm{th}}$ atomic site', xy=(0.3425, 5.24505181), xytext=(0.3725, 4.24505181), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))

plt.xlabel('Normalized length $r/W$')
plt.ylabel('Normalized stress $\sigma_{yy}/\sigma_{yy\ \mathrm{(boundary)}}$')
#plt.xlim([0.66,0.75])
plt.ylim([0, 15])
plt.rc('axes', titlesize=24) 
plt.rc('xtick', labelsize=24) 
plt.rc('ytick', labelsize=24) 
plt.rc('axes', labelsize=24)

#ax.legend()
plt.legend(loc = 'upper center',bbox_to_anchor=(0.5, 1.2), ncol=2, fancybox=True, shadow=False, prop={'size': 24})
plt.title('')
#grids
plt.minorticks_on()
#plt.grid (which='minor', color='g', ls='-.', lw=0.5)
#plt.grid(which='major',color='g', ls='--', lw=0.5)
#################for zooming particular range
axins = zoomed_inset_axes(ax, 3,loc= 10,borderpad=1) # zoom = 6
axins.plot(normalized_x,normalized_stress,'-ro')
axins.plot(pos_norm_604,sig_norm_604,'--kd')
axins.plot(pos_norm_5492,sig_norm_5492,'-.k*')
axins.plot(pos_norm_45588,sig_norm_45588,'--kv')
axins.annotate('5$^{\mathrm{th}}$ atomic site', xy=(0.345, 4.5849), xytext=(0.385, 4.5849), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.annotate('6$^{\mathrm{th}}$ atomic site', xy=(0.3475, 4.13029772), xytext=(0.3875, 4.13029772), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.annotate('7$^{\mathrm{th}}$ atomic site', xy=(0.35, 3.78644737), xytext=(0.39, 3.78644737), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.annotate('8$^{\mathrm{th}}$ atomic site', xy=(0.3525, 3.49558828), xytext=(0.3925, 3.49558828), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.annotate('9$^{\mathrm{th}}$ atomic site', xy=(0.355, 3.26207283), xytext=(0.395, 3.26207283), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.annotate('10$^{\mathrm{th}}$ atomic site', xy=(0.3575, 3.08520457), xytext=(0.3975, 2.78520457), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.annotate('11$^{\mathrm{th}}$ atomic site', xy=(0.36, 2.93615663), xytext=(0.40, 2.33615663), fontsize=24,
            arrowprops=dict(facecolor='gray',arrowstyle="->"))
axins.set_xlim(0.34,0.372) # Limit the region for zoom
axins.set_ylim(2.44,4.65)
mark_inset(ax, axins, loc1=2, loc2=4, fc="none", ec="0.")
axins.locator_params(axis='x',nbins=1)
axins.locator_params(axis='y',nbins=1)


#plt.show()
plt.savefig('fig_edge.png',dpi = 1200, bbox_inches='tight')
```
