```python
import matplotlib.pyplot as plt

# --- 1. Figure Setup ---
fig, ax = plt.subplots(figsize=(16, 9), facecolor='white')

# --- 2. Grayscale Style Definitions (12 Unique Combinations) ---
# Each tuple: (color, marker, linestyle, linewidth)
# We alternate black, dark gray, and light gray with different line patterns.
styles = [
    ('black',   'o', '-',  2.0),  # 1. Solid Black Circle
    ('#555555', 's', '-',  2.0),  # 2. Solid Dark Gray Square
    ('#999999', 'D', '-',  2.0),  # 3. Solid Light Gray Diamond
    ('black',   '^', '--', 2.0),  # 4. Dashed Black Triangle Up
    ('#555555', 'v', '--', 2.0),  # 5. Dashed Dark Gray Triangle Down
    ('#999999', 'p', '--', 2.0),  # 6. Dashed Light Gray Pentagon
    ('black',   '*', '-.', 2.0),  # 7. Dash-dot Black Star
    ('#555555', 'h', '-.', 2.0),  # 8. Dash-dot Dark Gray Hexagon
    ('#999999', '8', '-.', 2.0),  # 9. Dash-dot Light Gray Octagon
    ('black',   'X', ':',  2.0),  # 10. Dotted Black X
    ('#555555', 'P', ':',  2.0),  # 11. Dotted Dark Gray Plus (filled)
    ('#999999', 'd', ':',  2.0),  # 12. Dotted Light Gray Thin Diamond
]

# --- 3. Full Dataset List (12 Cases) ---
datasets = [
    (pn_50,  sn_50,  "crack depth $h/2$"),
    (pn_525, sn_525, "crack position $h/2.1$"),
    (pn_55,  sn_55,  "crack depth $h/2.22$"),
    (pn_575, sn_575, "crack position $h/2.35$"),
    (pn_60,  sn_60,  "crack depth $h/2.5$"),
    (pn_65,  sn_65,  "crack position $h/2.86$"),
    (pn_70,  sn_70,  "crack position $h/3.33$"),
    (pn_75,  sn_75,  "crack depth $h/4$"),
    (pn_80,  sn_80,  "crack depth $h/5$"),
    (pn_85,  sn_85,  "crack depth $h/6.67$"),
    (pn_90,  sn_90,  "crack position $h/10$"),
    (pn_95,  sn_95,  "crack depth $h/20$"),
]

# --- 4. Iterative Plotting ---
for (xdata, ydata, label), (color, marker, ls, lw) in zip(datasets, styles):
    ax.plot(xdata, ydata,
            linestyle=ls, color=color, marker=marker,
            linewidth=lw, markersize=10, 
            markerfacecolor=color,
            markeredgecolor='black' if color != 'black' else 'white', 
            markeredgewidth=0.8,
            label=label)

# --- 5. Annotations & Reference Lines ---
ax.axvline(x=0.67, linewidth=1.5, linestyle=':', color='black')
ax.axvline(x=0.69, linewidth=1.5, linestyle=':', color='black')

# Annotations
ax.annotate('crack tip', xy=(0.66, 0), xytext=(0.675, 1.2), fontsize=24,
            arrowprops=dict(facecolor='gray', shrink=0.05))
ax.annotate('$K$-dominated \n linear elastic \n region',
            xy=(0.68, 11), xytext=(0.68, 11), fontsize=24, ha='center')

# --- 6. Formatting & Legibility ---
ax.set_xlabel('Normalized length $x/W$', fontsize=24)
ax.set_ylabel('Normalized stresses $\sigma_{yy}/\sigma_{yy \ \mathrm{(boundary)}}$', fontsize=24)
ax.set_xlim([0.66, 0.75])
ax.set_ylim([0, 13])
ax.tick_params(axis='both', which='major', labelsize=24)
ax.minorticks_on()

# Adjusted legend to handle 12 grayscale items
ax.legend(loc='upper center', bbox_to_anchor=(0.5, 1.35), 
          ncol=3, fontsize=24, frameon=True, shadow=False)

plt.savefig('fig_12_cases_thesis.pdf', dpi=600, bbox_inches='tight')
plt.show()
```
